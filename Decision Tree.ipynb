```python
dataset=datasets.load_iris()
X = dataset.data
y = dataset.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
```

```python
model=DecisionTreeClassifier(criterion="entropy")
model.fit(X_train, y_train)

plt.figure(figsize=(12, 8))
plot_tree(model, filled=True, rounded=True)

plt.show()
```

```python
pred=model.predict(X_test)
confusion_matrix(y_test, pred)
```

```python
tree_classifier2 = DecisionTreeClassifier(criterion="gini", min_impurity_decrease=0.01, max_depth=4)
tree_classifier2.fit(X_train, y_train)

plt.figure(figsize=(12, 8))
plot_tree(tree_classifier2, filled=True, rounded=True)

plt.show()
```

```python
pred=tree_classifier2.predict(X_test)
confusion_matrix(y_test, pred)
```

```python
X, y = datasets.make_regression(n_samples=100, n_features=1, noise=20)

fig=plt.figure()
plt.grid()

plt.scatter(X,y)
plt.show()
```

```python
tree_classifier3 = DecisionTreeRegressor(criterion="squared_error", min_impurity_decrease=0.1, min_samples_leaf=8)

tree_classifier3.fit(X,y)

# Plot the decision tree
plt.figure(figsize=(12, 8))
plot_tree(tree_classifier3, filled=True, rounded=True)
plt.title("Decision Tree for Regression")
plt.show()
```

```python
fig=plt.figure()
xv=np.linspace(X.min(),X.max(), 1000).reshape(-1,1)
plt.grid()
pred=tree_classifier3.predict(xv)
plt.scatter(X,y)
plt.plot(xv, pred, 'r')
plt.title("Regression using Decision Tree")
plt.show()
```
